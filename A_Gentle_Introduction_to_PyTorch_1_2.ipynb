```
import pygame
import random

# Initialize Pygame
pygame.init()

# Screen dimensions
SCREEN_WIDTH = 400
SCREEN_HEIGHT = 600
SCREEN = pygame.display.set_mode((SCREEN_WIDTH, SCREEN_HEIGHT))
pygame.display.set_caption('Flappy Bird')

# Colors
WHITE = (255, 255, 255)
GREEN = (0, 255, 0)
BLUE = (0, 0, 255)
BLACK = (0, 0, 0)

# Game variables
FPS = 60
GRAVITY = 0.25
FLAP_STRENGTH = -5
PIPE_GAP = 150
PIPE_SPEED = 3

# Bird class
class Bird(pygame.sprite.Sprite):
    def __init__(self):
        super().__init__()
        self.image = pygame.Surface([30, 20])
        self.image.fill(BLUE) # Placeholder for bird image
        self.rect = self.image.get_rect(center=(50, SCREEN_HEIGHT // 2))
        self.velocity = 0

    def flap(self):
        self.velocity = FLAP_STRENGTH

    def update(self):
        self.velocity += GRAVITY
        self.rect.y += self.velocity

        # Keep bird on screen (for now, eventually game over)
        if self.rect.top < 0:
            self.rect.top = 0
            self.velocity = 0
        if self.rect.bottom > SCREEN_HEIGHT:
            self.rect.bottom = SCREEN_HEIGHT
            self.velocity = 0 # Game over condition in full game

# Pipe class
class Pipe(pygame.sprite.Sprite):
    def __init__(self, x_pos, y_pos, position):
        super().__init__()
        self.image = pygame.Surface([50, y_pos])
        self.image.fill(GREEN)
        if position == 'bottom':
            self.rect = self.image.get_rect(topleft=(x_pos, SCREEN_HEIGHT - y_pos))
        if position == 'top':
            self.rect = self.image.get_rect(bottomleft=(x_pos, y_pos))

    def update(self):
        self.rect.x -= PIPE_SPEED

# Game setup
bird_group = pygame.sprite.Group()
bird = Bird()
bird_group.add(bird)

pipes_group = pygame.sprite.Group()

# Function to create pipes
def create_pipe():
    random_pipe_pos = random.randint(150, SCREEN_HEIGHT - 150 - PIPE_GAP)
    bottom_pipe = Pipe(SCREEN_WIDTH, random_pipe_pos + PIPE_GAP, 'bottom')
    top_pipe = Pipe(SCREEN_WIDTH, random_pipe_pos, 'top')
    pipes_group.add(bottom_pipe)
    pipes_group.add(top_pipe)

# Timer for pipes
SPAWNPIPE = pygame.USEREVENT
pygame.time.set_timer(SPAWNPIPE, 1200) # Spawn a new pair of pipes every 1.2 seconds

# Game loop
clock = pygame.time.Clock()
running = True
game_active = True

while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
        if event.type == pygame.KEYDOWN:
            if event.key == pygame.K_SPACE and game_active:
                bird.flap()
        if event.type == SPAWNPIPE and game_active:
            create_pipe()

    SCREEN.fill(WHITE) # Background

    if game_active:
        # Update elements
        bird_group.update()
        pipes_group.update()

        # Draw elements
        bird_group.draw(SCREEN)
        pipes_group.draw(SCREEN)

        # Collision detection (simplified)
        if pygame.sprite.spritecollideany(bird, pipes_group):
            game_active = False # Game Over

        # Remove off-screen pipes
        for pipe in pipes_group.copy():
            if pipe.rect.right < 0:
                pipes_group.remove(pipe)
    else:
        # Display Game Over message
        font = pygame.font.Font(None, 50)
        game_over_text = font.render("GAME OVER! Press SPACE to Restart", True, BLACK)
        text_rect = game_over_text.get_rect(center=(SCREEN_WIDTH // 2, SCREEN_HEIGHT // 2))
        SCREEN.blit(game_over_text, text_rect)

        if event.type == pygame.KEYDOWN:
            if event.key == pygame.K_SPACE and not game_active:
                # Reset game
                bird.rect.center = (50, SCREEN_HEIGHT // 2)
                bird.velocity = 0
                pipes_group.empty()
                game_active = True


    pygame.display.update()
    clock.tick(FPS)

pygame.quit()
```
